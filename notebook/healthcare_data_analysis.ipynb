## Hospital Data Analysis

### Total Number of Patients

```python
query="select count(patient_id) as Total_Patients from patients;"
pd.read_sql(query,engine)
```

```python
query="select gender,count(*) as patients from patients group by gender;"
pd.read_sql(query,engine)
```

### Total Number of Doctors

```python
query="select count(doctor_id) as Total_Doctors from doctors;"
pd.read_sql(query,engine)
```

### Total Revenue

```python
query="select sum(amount) as Total_revenue from billing;"
pd.read_sql(query,engine)
```

### Total Number of Treatments

```python
query="select count(treatment_id) as Total_Treatments from treatments;"
pd.read_sql(query,engine)
```

### Highest Cost Treatment

```python
query="select treatment_type,cost from treatments where cost=(select max(cost) as Most_Expensive_treatments from treatments);"
pd.read_sql(query,engine)
```

### Lowest Cost Treatment

```python
query="select treatment_type,cost from treatments where cost=(select min(cost) as Most_Expensive_treatments from treatments);"
pd.read_sql(query,engine)
```

### Specialization Wise doctors

```python
query="select specialization,count(doctor_id) as Number_of_doctors from doctors group by specialization;"
pd.read_sql(query,engine)
```

### Branches of Hospital

```python
query="select hospital_branch,count(*) as count from doctors group by hospital_branch"
pd.read_sql(query,engine)
```

### Month Wise Number of Appointments

```python
query="select month(appointment_date) as month,count(*) number_of_appointments from appointments  group by month(appointment_date) order by month;"
pd.read_sql(query,engine)
```

### Doctor with Maximum Appointments

```python
query="select concat(d.first_name,' ',d.last_name) as Doctor,count(a.appointment_id) as appointments from doctors d inner join appointments a on d.doctor_id=a.doctor_id group by a.doctor_id order by appointments desc limit 1"
pd.read_sql(query,engine)
```

### Specialization with Appointment Numbers

```python
query="select d.specialization,count(a.appointment_id) as appointments from doctors d inner join appointments a on d.doctor_id=a.doctor_id group by d.specialization"
pd.read_sql(query,engine)
```

### Appointments Status

```python
query="select status,count(*) as total from appointments group by status;"
pd.read_sql(query,engine)
```

### Doctor with Highest Revenue

```python
query="select concat(d.first_name,' ',last_name) as Doctor,sum(t.cost) as Total_revenue from appointments a inner join doctors d on a.doctor_id=d.doctor_id inner join treatments t on a.appointment_id=t.appointment_id group by a.doctor_id order by Total_revenue desc limit 1;"
pd.read_sql(query,engine)
```

### Specialization wise Revenue

```python
query="select d.specialization,sum(t.cost) as Total_revenue from appointments a inner join doctors d on a.doctor_id=d.doctor_id inner join treatments t on a.appointment_id=t.appointment_id group by d.specialization;"
pd.read_sql(query,engine)
```

### Patient with Highest Average Bill Amount

```python
query="select concat(p.first_name,' ',p.last_name) as patient,AVG(b.amount) AS avg_amount FROM patients p INNER JOIN billing b ON p.patient_id = b.patient_id group by patient ORDER BY avg_amount DESC LIMIT 1;"
pd.read_sql(query,engine)
```

### Patient with Many Appoinments

```python
query="select concat(p.first_name,' ',p.last_name) as Patient,count(a.appointment_id)  as appointments from appointments a inner join patients p on a.patient_id=p.patient_id group by a.patient_id having count(a.appointment_id)= (select max(total_appointments) from (select count(appointment_id) as total_appointments from appointments group by patient_id) as appointment);"
pd.read_sql(query,engine)
```
